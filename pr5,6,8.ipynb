```python
#PRACTICAL:5
# text file
f1 = open("message.txt", "w")
f1.write("Welcome to Python programming\n")
f1.write("File Handling in Python")
f1.close()

f1 = open("message.txt", "r")
print("After Writing: ")
print(f1.read())
f1.close()

f1 = open("message.txt", "a")
f1.write("This is appended line\n")
f1.close()

# csv file
import csv

data = [["Id", "Students", "Score"],
        [101, "Pal", 90],
        [102, "Aarya", 85],
        [103, "Nancy", 76]]

f1 = open("marks.csv", "w", newline="")
writer = csv.writer(f1)
writer.writerows(data)
f1.close()

f1 = open("marks.csv", "r")
reader = csv.reader(f1)
for row in reader:
    print(row)
f1.close()

# binary file
bl = open("sample.bin", "wb")
bl.write(b"Python Binary Data")
bl.close()

bl = open("sample.bin", "rb")
print("After writting: ")
print(bl.read())
bl.close()

bl = open("sample.bin", "ab")
bl.write(b"\nAppended Data")
bl.close()
```

```python
#PRACTICAL:6
# try,expect,ese and fnally
try:
    n1 = 10
    n2 = 2
    r = n1 / n2
except ZeroDivisionError:
    print("Error: Can't Divide by zero")
else:
    print("Result: ", r)
finally:
    print("Execution completed")

# user define and raise
class AgeError(Exception):
    pass

try:
    age = 15
    if age < 18:
        raise AgeError("Age must be 18 or above")
except AgeError as c:
    print("Custom Exception: ", c)
else:
    print("You are eligible")
finally:
    print("Program Completed")
```

```python
#PRACTICAL:8
import sqlite3

# Connect to database
connection = sqlite3.connect("my_database.db")
cursor = connection.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INT PRIMARY KEY,
        name TEXT,
        age INT,
        course TEXT
    )
""")

sql = "INSERT OR IGNORE INTO students (id, name, age, course) VALUES (?, ?, ?, ?)"
values = [
    (1, "Pal", 20, "Python"),
    (2, "Aarya", 21, "Java"),
    (3, "Nancy", 19, "C++")
]

cursor.executemany(sql, values)
connection.commit()
print("Student inserted !!")

cursor.execute("SELECT * FROM students")
students = cursor.fetchall()
for student in students:
    print(student)

sql = "UPDATE students SET name = ?, age = ?, course = ? WHERE id = ?"
values = ("Nancy", 22, "MySQL", 3)
cursor.execute(sql, values)
connection.commit()
print("Student record updated !!")

cursor.execute("SELECT * FROM students")
students = cursor.fetchall()
for student in students:
    print(student)

sql = "DELETE FROM students WHERE id = ?"
student_id = (2,)
cursor.execute(sql, student_id)
connection.commit()
print("Student record deleted !!")

cursor.execute("SELECT * FROM students")
students = cursor.fetchall()
for student in students:
    print(student)

cursor.close()
connection.close()
```
