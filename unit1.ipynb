```python
# variable and data type
a = 20
b = 5.5
c = "Hello"
d = True

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))

# arithmetic operator
a = 18
b = 5

print("Addition: ", a + b)
print("Subtraction: ", a - b)
print("Multiplication: ", a * b)
print("Division: ", a / b)
print("Modulus: ", a % b)
print("Exponential: ", a ** b)
print("Floor Division: ", a // b)

# assignment operator
a = 20
print("a is: ", a)
a += 7
print("a += 7: ", a)
a -= 4
print("a -= 4: ", a)
a *= 3
print("a *= 3: ", a)
a /= 5
print("a /= 5: ", a)
a %= 5
print("a %= 5: ", a)

# comparison operator
a = 6
b = 4

print("a == b: ", a == b)
print("a >= b: ", a >= b)
print("a <= b: ", a <= b)
print("a > b: ", a > b)
print("a < b: ", a < b)
print("a != b: ", a != b)

# logical operator
a = 6
b = 4

print(a > 5 and b < 10)
print(a > 15 or b < 10)
print(not (a > 5))

# membership operator
fruits = ["apple", "mango"]

print("apple in fruits: ", "apple" in fruits)
print("Orange not in fruits: ", "Orange" not in fruits)
print("Kiwi in fruits: ", "kiwi" in fruits)

# identity operator
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print("a is b: ", a is b)
print("a is c: ", a is c)
print("a is not c: ", a is not c)

# bitwise operator
a = 10
b = 3

print("AND: ", a & b)
print("OR: ", a | b)
print("XOR: ", a ^ b)
print("Left Shift: ", a << 1)
print("Right Shift: ", a >> 1)
print("Not a: ", ~a)

l1 = [15, 25, 35, 25, 45]
l1.append(55)
l1.insert(1, 10)
l1.extend([65, 75])
l1.remove(25)
print("List: ", l1)
print("Pop: ", l1.pop())
print("Index of 35: ", l1.index(35))
print("Count of 25: ", l1.count(25))
l1.reverse()
print("Reversed: ", l1)
l1.sort()
print("Sorted: ", l1)
```

```python
# tuple
t1 = (15, 25, 35, 25, 45)
print("Tuple: ", t1)
print("Count of 25: ", t1.count(25))
print("Index of 35: ", t1.index(35))
print("Length: ", len(t1))
print("Slice: ", t1[1:4])

# set
s1 = {2, 4, 6, 8}
s2 = {6, 8, 10, 12}
s1.add(14)
print("Set 1: ", s1)
s1.remove(2)
print("Set 1: ", s1)
print("Union: ", s1.union(s2))
print("Intersect: ", s1.intersection(s2))
print("Difference: ", s1.difference(s2))
print("Symmetric Difference: ", s1.symmetric_difference(s2))
print("Subset: ", {4, 6}.issubset(s1))
print("Disjoint: ", s1.isdisjoint({20, 30}))

# dictionary
d1 = {"name": "Pal", "age": 20, "city": "Surat"}
print("Keys: ", list(d1.keys()))
print("Values: ", list(d1.values()))
print("Items: ", list(d1.items()))
print("Get age: ", d1.get("age"))
print("Pop city: ", d1.pop("city"))
d1.update({"branch": "AI"})
print("Dict: ", d1)

# string
s1 = "data science"
print("Upper: ", s1.upper())
print("Lower: ", s1.lower())
print("Title: ", s1.title())
print("Strip: ", s1.strip())
print("Replace: ", s1.replace("data", "computer"))
print("Find: ", s1.find("science"))
print("Split: ", s1.split())
print("Join: ", "-".join(["x", "y", "z"]))
```

```python
# if
age = 20
if age >= 18:
    print("You are eligible to vote")

# if else
num = 4
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")

# elif
marks = 75
if marks >= 80:
    print("Grade A")
elif marks < 80 and marks >= 60:
    print("Grade B")
elif marks < 60 and marks >= 33:
    print("Grade C")
else:
    print("Fail")

# while
i = 1
while i <= 5:
    print(i)
    i += 1

# for
for i in range(6):
    print(i)

# nested for
for i in range(2):
    for j in range(3):
        print(i, j)

# break
for i in range(6):
    if i == 3:
        break
    print(i)

# continue
for i in range(6):
    if i == 3:
        continue
    print(i)

# pass
for i in range(6):
    if i == 2:
        pass
    print(i)
```

```python
# without argument without return type
def greet():
    print("Hello world")

greet()

# with argument without return type
def greet(name):
    print("Hello", name)

greet("Pal")

# without argument with return type
def add():
    a = 6
    b = 4
    return a + b

sum_val = add()
print("Addition: ", sum_val)

# with argument with return type
def add(a, b):
    return a + b

sum_val = add(6, 4)
print("Addition: ", sum_val)

# default argument
def greet(name="Pal"):
    print("Hello", name)

greet()

# math,random and statistics function
import math
import random
import statistics

print("Square root of 49: ", math.sqrt(49))
print("Factorial of 6: ", math.factorial(6))
print("Random integer [10, 99]: ", random.randint(10, 99))
num_list = [10, 20, 30, 40, 50, 70]
print("Mean: ", statistics.mean(num_list))
```
