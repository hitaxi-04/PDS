```python
import pandas as pd

# Data Loading
df = pd.read_csv("/OnlineNewsPopularity.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Dataset loaded successfully")
print("Shape:", df.shape)

# 1. SLICING

print("\nFirst 5 rows:")
print(df.iloc[0:4])

# 2. FILTERING

filtered_data = df[df["shares"] > 1000]

print("\nArticles with more than 1000 shares:")
print(filtered_data[["n_tokens_title", "n_tokens_content", "shares"]].head())

# 3. SORTING

sorted_data = df.sort_values(by="shares", ascending=False)

print("\nTop 5 articles by shares:")
print(sorted_data[["n_tokens_title", "num_imgs", "num_videos", "shares"]].head())

# 4. CONCATENATING

data1 = df.iloc[:4]
data2 = df.iloc[:4]

combined_data = pd.concat([data1, data2])

print("\nConcatenated data:")
print(combined_data.head(5))

# 5. AGGREGATING

print("\nAggregation:")

print("Average shares:", df["shares"].mean())
print("Maximum shares:", df["shares"].max())
print("Minimum shares:", df["shares"].min())

print("\nAverage shares by number of images:")
print(df.groupby("num_imgs")["shares"].mean().head())

# 6. NORMALIZING

df["shares_normalized"] = (
    (df["shares"] - df["shares"].min()) /
    (df["shares"].max() - df["shares"].min())
)

print("\nNormalized shares:")
print(df[["shares", "shares_normalized"]].head())
```
